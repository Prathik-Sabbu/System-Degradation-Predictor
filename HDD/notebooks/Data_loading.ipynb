```python
import pandas as pd
import os

def show_dataset_columns():
    # Base directory for the data
    base_dir = "Data"
    
    # Dictionary mapping categories to sample files in your workspace
    datasets = {
        # "GPU": [
        #     "GPU/TitanGPULife-master/data/gc_full.csv",
        #     "GPU/TitanGPULife-master/data/gc_summary_loc.csv"
        # ],
        "HardDrive": [
            # Reading one file from Q1 as a representative sample for HardDrives
            "HardDrive/data_Q1_2025/data_Q1_2025/2025-01-01.csv"
        ]
        # "BSC": [
        #     "BSC/features.csv",
        #     "BSC/evaluations.csv",
        #     "BSC/predictions.csv",
        #     "BSC/ues_reduction.csv"
        # ],
        # "MIT": [
        #     "MIT/dcgm.csv",
        #     "MIT/scheduler_data.csv"
        # ],
        # "SSD": [
        #     "SSD/ssd_failure_tag.csv/ssd_failure_tag.csv",
        #     "SSD/location_info_of_ssd.csv/location_info_of_ssd.csv",
        #     "SSD/smart_log_20191231.csv/20191231.csv"
        # ]
    }

    for category, files in datasets.items():
        print(f"=======================================")
        print(f"       {category.upper()} DATA VALUES")
        print(f"=======================================")
        for file_path in files:
            full_path = os.path.join(base_dir, file_path)
            
            if os.path.exists(full_path):
                try:
                    # nrows=0 reads only the header to save memory and time
                    df = pd.read_csv(full_path, nrows=0)
                    columns = df.columns.tolist()
                    
                    print(f"File: {os.path.basename(file_path)}")
                    print(f"Data Values (Columns):")
                    # Formatting the list of columns nicely
                    for col in columns:
                        print(f"  - {col}")
                    print("\n")
                except Exception as e:
                    print(f"Error reading {file_path}: {e}\n")
            else:
                print(f"File not found: {full_path}\n")

if __name__ == "__main__":
    show_dataset_columns()
```
